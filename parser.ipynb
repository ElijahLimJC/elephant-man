```python
import pymupdf

reader = pymupdf.open("elephantman.pdf")
full_text = ""
for page in reader:
    full_text += page.get_text()
```

```python
count = 0
saved = []
previous_page_lines = []

for page_index, page in enumerate(reader):
    page_text = page.get_text()
    lines = page_text.splitlines()

    for idx, line in enumerate(lines):
        cleaned_line = line.strip()

        if "MERRICK :" in cleaned_line:
            if idx == 0 and previous_page_lines:
                saved.append(previous_page_lines[-1].strip())
            else:
                saved.append(lines[idx-1].strip())
            saved.append(cleaned_line)
            count += 2

    previous_page_lines = lines

print(count)

```

```python
with open("out.txt", "w") as file:
    # Join items with a newline character
    file.write("\n".join(saved))
```
